**1)A retail store wants to identify customers who make frequent purchases. 
Given the dataset below, write a Python program to:
Group customers by their IDs.**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
data = {'Customer_ID': [101, 102, 103, 101, 104, 102, 101, 105, 102, 103],
        'Purchase_Amount': [200, 150, 180, 220, 300, 200, 100, 400, 250, 300]}
```

```python
df = pd.DataFrame(data)
df
```

```python
grouped = df.groupby('Customer_ID')['Purchase_Amount'].sum()
grouped
```

```python
largest = df.groupby('Customer_ID')['Purchase_Amount'].sum().nlargest(3)
largest
```

**2)A real estate company wants to predict house prices based on square footage. Write a Python program to:
Train a Linear Regression model.**

```python
data1 = {'Square_Feet': [1500, 2000, 2500, 3000, 3500],
        'Price': [300000, 400000, 500000, 600000, 700000]}
```

```python
dataset = pd.DataFrame(data1)
dataset
```

```python
independent = dataset[["Square_Feet"]]
independent
```

```python
dependent = dataset[["Price"]]
dependent
```

```python
x_train,x_test,y_train,y_test = train_test_split(independent,dependent,test_size = 0.30, random_state = 0)
regressor = LinearRegression()
regressor.fit(x_train,y_train)
test_data = [[1800], [2800]]
y_pred = regressor.predict(test_data)
print("Prediction Price:")
print(y_pred)
```

**3)A company wants to identify the top 3 most common categories in a dataset.
Given the dataset below, write a Python program to: Group the data by Category.**

```python
data2 = {'Category': ['A', 'B', 'C', 'A', 'D', 'B', 'A', 'E', 'B', 'C', 'C', 'A'],
        'Value': [10, 15, 20, 30, 25, 18, 22, 40, 35, 50, 45, 15]}
```

```python
df2 = pd.DataFrame(data2)
df2
```

```python
grouped = df2.groupby('Category')['Value'].count()
print("Total Occurrences per Category:")
grouped
```

```python
largest = df2.groupby('Category')['Value'].count().nlargest(3)
print("Top 3 Frequent Categories:")
largest
```

**4)A dataset contains missing values in the Age column. Write a Python program to:
Replace missing values with the mean of the column.**

```python
data3 = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'Age': [25, 30, None, 35, None]}
```

**5)Implementing a Simple Linear Regression Model
You are given a dataset with Experience (years) and Salary ($). Write a Python program to:
Train a Linear Regression model.**

```python
data4 = {'Experience': [1, 2, 3, 4, 5], 
        'Salary': [30000, 35000, 40000, 45000, 50000]}
```

```python
df4 = pd.DataFrame(data4)
df4
```

```python
independent1 = df4[["Experience"]]
independent1
```

```python
dependent1 = df4[["Salary"]]
dependent1
```

```python
x_train,x_test,y_train,y_test = train_test_split(independent1,dependent1,test_size = 0.30, random_state = 0)
model = LinearRegression()
model.fit(x_train,y_train)
test_data = [[6]]
result = model.predict(test_data)
print("Predicted Salary for 6 years of experience:")
print(result)
```
